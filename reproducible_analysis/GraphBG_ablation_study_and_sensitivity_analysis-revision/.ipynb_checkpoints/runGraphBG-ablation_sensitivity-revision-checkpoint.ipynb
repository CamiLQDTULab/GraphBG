### Ablation

```python
## Read data
#data_dir = 'Data/151507.h5ad'
for data_name in dataname_list:
    print(data_name)
    #data_dir = '/data/hoan/spatial_transcriptomics/data/SDMBench/Data/Maynard/151507.h5ad'
    data_dir = '/data/hoan/spatial_transcriptomics/data/SDMBench/Data/Maynard/'+data_name+'.h5ad'
    data_name = data_dir[-11:-5]
    adata=sc.read_h5ad(data_dir)
    
    ## Preproc data
    n_features = adata.shape[1]
    if n_features > 5000:
        sc.pp.highly_variable_genes(adata, flavor="seurat_v3", n_top_genes=3000)
    sc.pp.normalize_total(adata, target_sum=1e4)
    sc.pp.log1p(adata)
    sc.pp.scale(adata, zero_center=False, max_value=10)
    if n_features > 5000:
        adata =  adata[:, adata.var['highly_variable']]
    
    ## Perform PCA
    sc.tl.pca(adata, n_comps=20)
    
    ## Run GraphBG ablation
    from utils import runGraphBG
    from utils import runGraphBG_PCAonly
    
    n_clusters = 7
    #runGraphBG_PCAonly(adata, n_clusters=n_clusters, radius=50, key='X_pca', refinement=True, n_neighbors = 4, n_components = 20,
    # covariance_type = 'tied', n_init = 10, max_iter = 1000)
    runGraphBG(adata, n_clusters=n_clusters, radius=50, key='X_pca', refinement=False, n_neighbors = 4, n_components = 20,
                                                                covariance_type = 'tied', n_init = 10, max_iter = 1000)
    GraphBG_labels = adata.obs['domain'].values
    ## Compute the NMI with ground truth
    true_labels = pd.factorize(adata.obs['Region'])[0]
    print(normalized_mutual_info_score(true_labels[true_labels>=0], GraphBG_labels[true_labels>=0]))
    
    ## Save output
    df = pd.DataFrame({'ground_truth': true_labels[true_labels>=0], 'cluster': GraphBG_labels[true_labels>=0]})
    # df.to_csv('output/ablation/GraphBG_PCAonly_'+data_name+'.csv',index=None)
    #df.to_csv('output/ablation/GraphBG_PCAandASGC_'+data_name+'.csv',index=None)
    df.to_csv('output/ablation/GraphBG_full_'+data_name+'.csv',index=None)
```

#### number of PCs

```python
## Read data
#data_dir = 'Data/151507.h5ad'
for data_name in dataname_list:
    print(data_name)
    #data_dir = '/data/hoan/spatial_transcriptomics/data/SDMBench/Data/Maynard/151507.h5ad'
    data_dir = '/data/hoan/spatial_transcriptomics/data/SDMBench/Data/Maynard/'+data_name+'.h5ad'
    data_name = data_dir[-11:-5]
    adata=sc.read_h5ad(data_dir)
    
    ## Preproc data
    n_features = adata.shape[1]
    if n_features > 5000:
        sc.pp.highly_variable_genes(adata, flavor="seurat_v3", n_top_genes=3000) #default = 3000
    sc.pp.normalize_total(adata, target_sum=1e4)
    sc.pp.log1p(adata)
    sc.pp.scale(adata, zero_center=False, max_value=10)
    if n_features > 5000:
        adata =  adata[:, adata.var['highly_variable']]
    
    ## Perform PCA
    sc.tl.pca(adata, n_comps=100)
    
    ## Run GraphBG ablation
    from utils import runGraphBG    
    n_clusters = 7
    for n_PCA in [10, 20, 30, 50, 80, 100]:
        runGraphBG(adata, n_clusters=n_clusters, radius=50, key='X_pca', refinement=True, n_neighbors = 4, n_components = n_PCA,
                                                                    covariance_type = 'tied', n_init = 10, max_iter = 1000)
        GraphBG_labels = adata.obs['domain'].values
        ## Compute the NMI with ground truth
        true_labels = pd.factorize(adata.obs['Region'])[0]
        print(normalized_mutual_info_score(true_labels[true_labels>=0], GraphBG_labels[true_labels>=0]))
        
        ## Save output
        df = pd.DataFrame({'ground_truth': true_labels[true_labels>=0], 'cluster': GraphBG_labels[true_labels>=0]})
        # df.to_csv('output/ablation/GraphBG_PCAonly_'+data_name+'.csv',index=None)
        df.to_csv('output/ablation/GraphBG_nPCA_' +str(n_PCA)+'_'+data_name+'.csv',index=None)
```
